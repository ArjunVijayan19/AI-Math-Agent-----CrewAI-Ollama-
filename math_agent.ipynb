```python
from crewai import Agent,Task,Crew,LLM
from crewai.tools import tool
```

```python
local_LLM=LLM(
    model='ollama/llama3.2:1b',
    base_url="http://localhost:11434"
)
```

```python
@tool
def calculator(expression:str) ->str:
    "Evaluate a athematical expression"
    try:
        return str(eval(expression))
    except:
        return 'Error in Calculation'
```

```python
math_agent=Agent(
    role='Math Solver',
    goal="Solve mathematical expressions using the claculator tool",
    backstory="You are a smart mathematician who use tools to solve problem accurately",
    llm=local_LLM,
    tools=[calculator],
    verbose=True

)
```

```python
task=Task(
    description="Calculate the result of (45+55)*5",
    expected_output="Correct Numerical Answer",
    agent=math_agent
)
```

```python
crew=Crew(
    agents=[math_agent],
    tasks=[task],
    verbose=True
)
```

```python
result=crew.kickoff()
print("\n Final Answer")
print(result)
```
